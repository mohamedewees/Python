Lists 

```python
myList = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
```

```python
print(myList)
print(myList[-1])  # last element in the array
print(myList[-2])  # last second element
print(myList[2:5]) # print range of elements in the array
print(myList[:4])  # print elements from the beginning of the array till element number 4 exclussive
print(myList[2:])  # print elements from element number 2 inclussive till the end of the array
```

```python
thislist = ["apple", "banana", "cherry"]
thislist.insert(2, "watermelon")  # we must identify the location for the new 
print(thislist)
```

```python
list2 = ["apple", "banana", "cherry"]
list2.append("orange")
print(list2)
```

```python
# To append elements from another list to the current list
list3 = ["apple", "banana", "cherry"]
list4 = ["mango", "pineapple", "papaya"]
list3.extend(list4)
print(list3)
```

```python
myList = ["apple", "banana", "cherry"]
myList.remove("apple")  # to remove element
print(myList)
```

```python
myList =  ["apple", "banana", "cherry"]
myList.clear()  # clear the list 
print(myList)
```

```python
# loop on a list
myList = ["apple", "banana", "cherry"]
for x in myList:
    print(x)
```

```python
# loop on the index in a list
myList = ["apple", "banana", "cherry"]
for i in range(len(myList)):
    print(myList[i])
```

```python
# list comprehension enables you to loop on a list in one line
myList = ["apple", "banana", "cherry"]
[print(x) for x in myList]
```

```python
# another example
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newList = []
for x in fruits:
    if 'a' in x:
        newList.append(x)
print(newList)

###############################
# let's do it with list comperehnsion
# syntax: newlist = [expression for item in iterable if condition == True]
fruits1 = ["apple", "banana", "cherry", "kiwi", "mango"]
newList1 = [x for x in fruits1 if 'a' in x]
print(newList1)
```

```python
# sort list (by default sort method sorting the list alphapitically ascending)
myList = ["orange", "mango", "kiwi", "pineapple", "banana"]
myList.sort()
print(myList)
```

```python
myList = [100, 50, 65, 82, 23]
myList.sort()
print(myList)
```

```python
# sort descending
myList = [100, 50, 65, 82, 23]
myList.sort(reverse=True)
print(myList)
```

```python
# Tuple
# A tuple is a collection which is ordered and unchangeable.
myTuple = ("apple", "banana", "cherry")
print(myTuple)
```

```python
# same access of List applies for Tuple
myTuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(myTuple[-1])  # last element in the array
print(myTuple[-2])  # last second element
print(myTuple[2:5]) # print range of elements in the array
print(myTuple[:4])  # print elements from the beginning of the array till element number 4 exclussive
print(myTuple[2:])  # print elements from element number 2 inclussive till the end of the array
```

```python
# to update or change anyvalue in a tuple we have to convert it first to a list
# same applies if we want to add or remove item from a tuple
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "Kiwi"
x = tuple(y)
print(x)
```

```python
# Dictionaries
# Dictionaries are used to store data values in key:value pairs.
myDict = {"brand":"ford",
          "model":"Mostang",
          "Year":1964}
print(myDict)
# we can print value by calling it's related key
print(myDict["brand"])
```

```python
# we can use also get methon to retreive a value by it's key
myDict={
    "brand":"ford",
    "model":"Mostang",
    "Year":1964
        }
x = myDict.get("brand")
print(x)
```

```python
# get list of keys in a dictionary
x = myDict.keys()
print(x)
```

```python
# get list of values in a dictionary
x = myDict.values()
print(x)
```

```python
# get a list of key value pair
x = myDict.items()
print(x)
```

```python
# change item in a dictionary
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict["year"] = 1990
print(thisdict.items())
```

```python
# add item
thisdict["color"] = "red"
print(thisdict)
```

```python
# loop on dictionary:
for x in thisdict:
    print(x)  # prints the keys
    print(thisdict[x])  # prints the values
```

```python
# loop on the items:
for x,y in thisdict.items():
    print(x,y)
```
