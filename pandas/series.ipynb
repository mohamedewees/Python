```python
import pandas as pd
```

```python
# series is one dimensional array
age = pd.Series([11,22,33,44])
print(age)
```

```python
# if we added multiple datatypes to the series it will be obeject
age = pd.Series([11,"Hello",True])
age
```

```python
# we can change the index as well
age = pd.Series([11,22,33,44],index=['A','B','C','D'])
age
```

```python
#data frame like excell sheet (table of rows and columns)
studentInfo = pd.DataFrame({
    "Name": ["Mohamed","Nada","Celine","Malek"],
    "Age": [36,34,3,1],
    "isGraduated": [True,True,False,False]
})
studentInfo
```

```python
# we can do the same for the index
studentInfo = pd.DataFrame({
    "Name": ["Mohamed","Nada","Celine","Malek"],
    "Age": [36,34,3,1],
    "isGraduated": [True,True,False,False]
    }, index=['A','B','C','D'])
studentInfo
```
